```python
df["Risk_Month_to_Month"] = (df["Contract"] == "Month-to-month").astype(int)

df["Risk_Month_to_Month"].value_counts()
```

```python
df["Risk_Fiber_Optic"] = (
    df["InternetService"] == "Fiber optic"
).astype(int)

df["Risk_Fiber_Optic"].value_counts()
```

```python
df["Risk_Electronic_Check"] = (
    df["PaymentMethod"] == "Electronic check"
).astype(int)

df["Risk_Electronic_Check"].value_counts()
```

```python
df["Risk_No_TechSupport"] = (
    df["TechSupport"] == "No"
).astype(int)

df["Risk_No_TechSupport"].value_counts()
```

```python
df["Risk_No_OnlineSecurity"] = (
    df["OnlineSecurity"] == "No"
).astype(int)

df["Risk_No_OnlineSecurity"].value_counts()
```

```python
df["Risk_New_Customer"] = (
    df["tenure"] <= 12
).astype(int)

df["Risk_New_Customer"].value_counts()
```

```python
df["Risk_Score"] = (
    df["Risk_Month_to_Month"]
    + df["Risk_Fiber_Optic"]
    + df["Risk_Electronic_Check"]
    + df["Risk_No_TechSupport"]
    + df["Risk_No_OnlineSecurity"]
    + df["Risk_New_Customer"]
)
```

```python
df["Risk_Score"].value_counts().sort_index()
```

```python
risk_churn = (
    df.groupby("Risk_Score", observed=True)["Churn"]
      .apply(lambda x: (x == "Yes").mean() * 100)
      .reset_index(name="Churn_Rate")
)

risk_churn
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.plot(
    risk_churn["Risk_Score"],
    risk_churn["Churn_Rate"],
    marker="o"
)

plt.title("Churn Rate by Customer Risk Score")
plt.xlabel("Customer Risk Score")
plt.ylabel("Churn Rate (%)")

plt.xticks(risk_churn["Risk_Score"])

plt.show()
```

```python
def risk_category(score):
    if score <= 1:
        return "Low Risk"
    elif score <= 3:
        return "Medium Risk"
    else:
        return "High Risk"

df["Risk_Category"] = df["Risk_Score"].apply(risk_category)
```

```python
risk_summary = (
    df.groupby("Risk_Category", observed=True)
      .agg(
          Customers=("customerID", "count"),
          Monthly_Revenue=("MonthlyCharges", "sum"),
          Churn_Rate=("Churn", lambda x: (x == "Yes").mean() * 100)
      )
      .reset_index()
)

risk_summary
```

```python
import matplotlib.pyplot as plt

risk_order = ["Low Risk", "Medium Risk", "High Risk"]

plot_data = risk_summary.set_index("Risk_Category").loc[risk_order]

plt.figure(figsize=(8, 5))

plt.bar(
    plot_data.index,
    plot_data["Churn_Rate"]
)

plt.title("Churn Rate by Customer Risk Category")
plt.xlabel("Risk Category")
plt.ylabel("Churn Rate (%)")

plt.show()
```
